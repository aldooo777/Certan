## Outliers


```python
path = "weight-height_outlier.csv"
df2 = pd.read_csv(path)
df2
```

```python
import seaborn as sns
sns.boxplot(df2['Height'])
```

```python
import numpy as np
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Height'], [25, 75])
# Hitung IQR
iqr = q3 - q1
# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Height'] >= lower_bound)
 & (df2['Height'] <= upper_bound)]
sns.boxplot(clean_data['Height'])
```

```python
sns.boxplot(df2['Weight'])
```

```python
# Identifikasi Q1 Q3
q1, q3 = np.percentile(df2['Weight'], [25, 75])
# Hitung IQR
iqr = q3 - q1
# Cari lower dan upper bounds
lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)
# Hapus outliers
clean_data = df2[(df2['Weight'] >= lower_bound)
 & (df2['Weight'] <= upper_bound)]
sns.boxplot(clean_data['Weight'])
```

```python
import matplotlib.pyplot as plt

# Membuat daftar kolom yang ingin dianalisis, yaitu "Weight" dan "Height".
columns_to_plot = ['Weight', 'Height']

# Memastikan kolom ada di dataset
columns_to_plot = [col for col in columns_to_plot if col in df2.columns]

fig, axs = plt.subplots(
    len(columns_to_plot),
    1,
    dpi=95,
    figsize=(7, 5 * len(columns_to_plot))
)

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(df2[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
def remove_outliers_sequentially(df2, columns):
    df_cleaned = df2.copy()
    for col in columns:
        q1, q3 = np.percentile(df_cleaned[col], [25, 75])
        iqr = q3 - q1
        lower_bound = q1 - (1.5 * iqr)
        upper_bound = q3 + (1.5 * iqr)
        df_cleaned = df_cleaned[(df_cleaned[col] >= lower_bound) & (df_cleaned[col] <= upper_bound)]
    return df_cleaned
# Terapkan fungsi untuk menghapus outliers dari "Height" dan "Weight"
clean_data = remove_outliers_sequentially(df2, ['Height', 'Weight'])

# Plot boxplots setelah menghapus outliers
columns_to_plot = ['Weight', 'Height']
fig, axs = plt.subplots(len(columns_to_plot), 1, dpi=95, figsize=(7, 5 * len(columns_to_plot)))

if len(columns_to_plot) == 1:
    axs = [axs]

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(clean_data[col], vert=False)
    axs[i].set_ylabel(col)

plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0, 1))

# Memisahkan kolom numerik dan non-numerik
numeric_cols = clean_data.select_dtypes(include=[np.number]).columns  # Ambil kolom numerik
non_numeric_cols = clean_data.select_dtypes(exclude=[np.number]).columns  # Ambil kolom non-numerik

rescaled_numeric = scaler.fit_transform(clean_data[numeric_cols])

rescaled_numeric[:]
```

```python
# Konversi kembali hasil scaling ke DataFrame
rescaled_df = pd.DataFrame(rescaled_numeric, columns=numeric_cols, index=clean_data.index)

# Gabungkan kembali dengan kolom non-numerik tanpa mengubah jumlah baris
final_df = pd.concat([clean_data[non_numeric_cols], rescaled_df], axis=1)

final_df
```

#### TUGAS NO 2
Handling missing value dari dataset 'Titanic-Dataset.csv'



Pada tugas ini dilakukan proses handling missing value pada dataset `Titanic-Dataset.csv`. Tahapan yang dilakukan meliputi membaca dataset, mengecek jumlah missing value pada setiap kolom, menangani nilai yang kosong, kemudian melakukan pengecekan kembali untuk memastikan seluruh missing value telah berhasil ditangani.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('Titanic-Dataset.csv')
df.head()
```

Dataset `Titanic-Dataset.csv` dibaca menggunakan library Pandas dan disimpan ke dalam DataFrame `df`. Fungsi `df.head()` digunakan untuk menampilkan lima baris pertama sehingga dapat diketahui gambaran awal mengenai struktur dan isi dataset yang akan digunakan.

Pada tahap ini dilakukan pemeriksaan terhadap missing value pada setiap kolom menggunakan `df.isnull().sum()`. Pemeriksaan ini bertujuan untuk mengetahui kolom mana yang memiliki data kosong serta mengetahui jumlah missing value pada masing-masing kolom sebelum dilakukan proses penanganan.

```python
df.isnull().sum()
```

Missing value pada dataset ditangani berdasarkan tipe datanya. Untuk kolom numerik, nilai yang kosong diisi menggunakan nilai median dari kolom tersebut. Median digunakan karena relatif tidak terpengaruh oleh nilai ekstrem. Sementara itu, untuk kolom non-numerik atau kategorikal, missing value diisi menggunakan modus, yaitu nilai atau kategori yang paling sering muncul dalam kolom tersebut.

```python
# Menangani missing value pada kolom numerik
numeric_cols = df.select_dtypes(include=[np.number]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Menangani missing value pada kolom non-numerik
categorical_cols = df.select_dtypes(exclude=[np.number]).columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

Setelah proses penanganan dilakukan, pemeriksaan `df.isnull().sum()` dilakukan kembali untuk memastikan bahwa tidak ada lagi missing value pada dataset. Berdasarkan hasil pemeriksaan, seluruh kolom memiliki jumlah missing value sebesar 0. Dengan demikian, seluruh missing value pada dataset Titanic telah berhasil ditangani.

```python
# Mengecek kembali missing value setelah ditangani
df.isnull().sum()
```

### Kesimpulan

Berdasarkan proses yang telah dilakukan pada dataset `Titanic-Dataset.csv`, seluruh missing value berhasil ditangani. Pada kolom numerik, missing value diisi menggunakan nilai median, sedangkan pada kolom non-numerik atau kategorikal, missing value diisi menggunakan modus. Setelah proses penanganan selesai, dilakukan pengecekan kembali menggunakan `df.isnull().sum()`. Hasil pemeriksaan menunjukkan bahwa seluruh kolom memiliki jumlah missing value sebesar 0. Dengan demikian, dataset Titanic sudah tidak memiliki missing value dan siap digunakan untuk proses analisis selanjutnya.


#### TUGAS NO 3
Handling outliers dari dataset 'diabetes.csv'


Pada tugas ini dilakukan proses deteksi dan handling outliers pada dataset `diabetes.csv`. Outlier merupakan nilai yang memiliki karakteristik berbeda atau berada jauh dari pola umum data. Untuk mendeteksi outlier, dipilih tiga kolom numerik yaitu `Glucose`, `BloodPressure`, dan `BMI`. Proses deteksi dilakukan menggunakan visualisasi boxplot dan metode Interquartile Range (IQR).

```python
# Membaca dataset diabetes
diabetes = pd.read_csv('diabetes.csv')

# Menampilkan 5 data pertama
diabetes.head()
```

Dataset `diabetes.csv` dibaca menggunakan library Pandas dan disimpan ke dalam DataFrame `diabetes`. Fungsi `diabetes.head()` digunakan untuk menampilkan lima baris pertama dataset sehingga dapat diketahui struktur dan contoh data yang akan dianalisis.

```python
# Melihat informasi dataset
diabetes.info() 
```

Fungsi `diabetes.info()` digunakan untuk mengetahui informasi mengenai dataset, seperti jumlah baris, jumlah kolom, nama kolom, jumlah data yang tidak kosong, dan tipe data setiap kolom. Berdasarkan hasil pemeriksaan, dataset diabetes memiliki 768 baris dan 9 kolom.

Untuk mendeteksi outlier, dipilih tiga kolom yaitu `Glucose`, `BloodPressure`, dan `BMI`. Ketiga kolom tersebut merupakan data numerik sehingga dapat divisualisasikan menggunakan boxplot. Boxplot digunakan untuk melihat distribusi data dan mengidentifikasi nilai yang berada di luar batas whisker yang dapat menjadi indikasi adanya outlier.

```python
import matplotlib.pyplot as plt

# Memilih 3 kolom untuk deteksi outlier
columns_to_plot = ['Glucose', 'BloodPressure', 'BMI']

# Membuat boxplot
fig, axs = plt.subplots(
    len(columns_to_plot),
    1,
    dpi=95,
    figsize=(7, 12)
)

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(diabetes[col], vert=False)
    axs[i].set_xlabel(col)
    axs[i].set_title(f'Boxplot {col} Sebelum Outlier Ditangani')

plt.tight_layout()
plt.show()
```

Setelah melakukan deteksi secara visual menggunakan boxplot, outlier dihitung secara matematis menggunakan metode Interquartile Range (IQR). Nilai IQR diperoleh dari selisih antara kuartil ketiga (Q3) dan kuartil pertama (Q1). Batas bawah ditentukan dengan rumus Q1 − 1,5 × IQR, sedangkan batas atas ditentukan dengan rumus Q3 + 1,5 × IQR. Data yang berada di luar batas tersebut dikategorikan sebagai outlier.

```python
# Menghitung outlier dengan metode IQR

for col in columns_to_plot:
    Q1 = diabetes[col].quantile(0.25)
    Q3 = diabetes[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = diabetes[
        (diabetes[col] < lower_bound) |
        (diabetes[col] > upper_bound)
    ]

    print(f"Kolom: {col}")
    print(f"Q1           : {Q1}")
    print(f"Q3           : {Q3}")
    print(f"IQR          : {IQR}")
    print(f"Lower Bound  : {lower_bound}")
    print(f"Upper Bound  : {upper_bound}")
    print(f"Jumlah Outlier: {len(outliers)}")
    print("-" * 40)
```

Berdasarkan perhitungan menggunakan metode IQR, ditemukan 5 outlier pada kolom `Glucose`, 45 outlier pada kolom `BloodPressure`, dan 19 outlier pada kolom `BMI`. Hasil tersebut menunjukkan bahwa ketiga kolom memiliki nilai yang berada di luar batas normal berdasarkan metode IQR.

Outlier ditangani dengan menghapus baris data yang memiliki nilai di luar batas bawah dan batas atas berdasarkan metode IQR. Dataset asli disalin terlebih dahulu ke dalam variabel `diabetes_clean` agar dataset asli tetap tersimpan dan tidak mengalami perubahan secara langsung.

```python
# Membuat salinan dataset
diabetes_clean = diabetes.copy()

# Menghapus outlier menggunakan metode IQR
for col in columns_to_plot:
    Q1 = diabetes_clean[col].quantile(0.25)
    Q3 = diabetes_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    diabetes_clean = diabetes_clean[
        (diabetes_clean[col] >= lower_bound) &
        (diabetes_clean[col] <= upper_bound)
    ]

# Menampilkan jumlah data
print("Jumlah data sebelum :", len(diabetes))
print("Jumlah data sesudah :", len(diabetes_clean))
print("Jumlah data dihapus :", len(diabetes) - len(diabetes_clean))
```

Dataset diabetes pada awalnya memiliki 768 baris data. Setelah dilakukan proses handling outlier secara berurutan pada kolom `Glucose`, `BloodPressure`, dan `BMI`, jumlah data menjadi 709 baris. Dengan demikian, sebanyak 59 baris data dihapus karena termasuk dalam data yang berada di luar batas IQR.

Setelah proses handling outlier selesai dilakukan, boxplot dibuat kembali menggunakan dataset `diabetes_clean`. Visualisasi ini digunakan untuk melihat perubahan distribusi data setelah nilai-nilai yang dianggap sebagai outlier berdasarkan metode IQR dihapus.

```python
# Boxplot setelah outlier ditangani

fig, axs = plt.subplots(
    len(columns_to_plot),
    1,
    dpi=95,
    figsize=(7, 12)
)

for i, col in enumerate(columns_to_plot):
    axs[i].boxplot(diabetes_clean[col], orientation='horizontal')
    axs[i].set_xlabel(col)
    axs[i].set_title(f'Boxplot {col} Setelah Outlier Ditangani')

plt.tight_layout()
plt.show()
```

### Kesimpulan

Berdasarkan proses yang telah dilakukan, outlier pada dataset `diabetes.csv` berhasil dideteksi dan ditangani. Deteksi dilakukan pada tiga kolom, yaitu `Glucose`, `BloodPressure`, dan `BMI`, menggunakan boxplot dan metode Interquartile Range (IQR). Ditemukan 5 outlier pada `Glucose`, 45 outlier pada `BloodPressure`, dan 19 outlier pada `BMI`. Setelah dilakukan handling dengan menghapus data yang berada di luar batas IQR, jumlah data berkurang dari 768 menjadi 709 baris. Boxplot setelah proses handling menunjukkan bahwa nilai-nilai ekstrem telah berkurang.
