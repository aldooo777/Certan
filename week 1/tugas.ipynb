```python
# Tugas 1
# Membuat matriks 5x5 dengan elemen random

import numpy as np

matriks = np.random.randint(1, 100, size=(5, 5))

print("Matriks 5x5:")
print(matriks)

print("\nNilai maksimum =", matriks.max())
print("Nilai minimum   =", matriks.min())
print("Nilai rata-rata =", matriks.mean())
```

```python
# Tugas 2
# Grafik 2 dimensi

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)

X, Y = np.meshgrid(x, y)

F = (X**6
     + 3*X**4*Y**2
     - 3*X**4
     + 3*X**2*Y**4
     - X**2*Y**3
     - 6*X**2*Y**2
     + 3*X**2
     + Y**6
     - 3*Y**4
     + 3*Y**2
     - 1)

plt.figure(figsize=(8, 8))

plt.contour(X, Y, F, levels=[0])

plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2 Dimensi')
plt.grid(True)
plt.axis('equal')

plt.show()
```

```python
# Tugas 3
# Grafik 3 dimensi

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = ((5*Y/4) - np.abs(X))**2 + X**2

fig = plt.figure()

ax = plt.axes(projection='3d')

ax.contour3D(X, Y, Z, 50, cmap='binary')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
```
