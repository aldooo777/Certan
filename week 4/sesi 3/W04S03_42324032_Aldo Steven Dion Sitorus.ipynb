## Naïve Bayes

In [1] — Import Library

Pada bagian ini dilakukan proses import beberapa library Python yang diperlukan untuk praktikum. NumPy digunakan untuk pengolahan data numerik, Pandas digunakan untuk mengolah data dalam bentuk DataFrame, sedangkan load_iris digunakan untuk mengambil dataset Iris dari Scikit-Learn.

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

In [2] — Konversi Dataset Menjadi DataFrame

Pada bagian ini dataset Iris dikonversi menjadi DataFrame menggunakan Pandas agar data lebih mudah diolah. Kolom Species ditambahkan sebagai label atau target berdasarkan data dari dataset Iris. Variabel X berisi data fitur, sedangkan variabel y berisi label species.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

In [3] — Menampilkan Label Species

Pada bagian ini variabel y ditampilkan untuk melihat label atau target dari dataset Iris. Label tersebut terdiri dari tiga jenis species, yaitu setosa, versicolor, dan virginica.

```python
y
```

In [4] — Mengubah Label Menjadi Angka

Pada bagian ini digunakan LabelEncoder untuk mengubah label species yang berbentuk teks menjadi angka. setosa akan direpresentasikan dengan angka 0, versicolor dengan angka 1, dan virginica dengan angka 2. Pengubahan ini dilakukan agar label dapat digunakan dalam proses klasifikasi oleh model Naïve Bayes.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

In [5] — Membagi Dataset Menjadi Data Training dan Testing

Pada bagian ini dataset dibagi menjadi data training dan data testing menggunakan train_test_split. Sebanyak 80% data digunakan untuk training, sedangkan 20% digunakan untuk testing. Parameter random_state=0 digunakan agar pembagian data tetap sama ketika kode dijalankan kembali.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

In [6] — Membuat dan Melatih Model Gaussian Naïve Bayes

Pada bagian ini dibuat model menggunakan algoritma Gaussian Naïve Bayes. Algoritma ini digunakan karena fitur pada dataset Iris berupa data numerik kontinu. Setelah model dibuat menggunakan GaussianNB(), model dilatih menggunakan data training melalui fungsi fit().

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

In [7] — Melakukan Prediksi Data Testing

Pada bagian ini model Gaussian Naïve Bayes digunakan untuk melakukan prediksi terhadap data testing. Fungsi predict() menghasilkan kelas prediksi untuk setiap data yang terdapat pada X_test.

```python
result = NB_model.predict(X_test)
result
```

In [8] — Menghitung Akurasi Model

Pada bagian ini digunakan accuracy_score untuk menghitung tingkat ketepatan model. Akurasi diperoleh dengan membandingkan hasil prediksi model dengan label sebenarnya dari data testing.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

In [9] — Prediksi Data Baru

Pada bagian ini dilakukan prediksi terhadap satu data baru. Data yang digunakan memiliki nilai sepal length, sepal width, petal length, dan petal width. Hasil prediksi dari model masih berupa angka sehingga digunakan inverse_transform() untuk mengembalikannya menjadi nama species.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

In [10] — Menggunakan DataFrame untuk Data Baru

Pada bagian sebelumnya muncul warning karena model dilatih menggunakan DataFrame yang memiliki nama kolom, sedangkan data baru diberikan dalam bentuk array tanpa nama kolom. Oleh karena itu, pada bagian ini data baru diubah menjadi DataFrame dan diberikan nama kolom yang sama dengan dataset Iris agar format data konsisten dengan data yang digunakan ketika model dilatih.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
